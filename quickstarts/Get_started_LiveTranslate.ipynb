Exercise 1

Variables, data types, input and output

```
name = input("Enter your name: ")
age_str = input("Enter your age: ")
age = int(age_str)

print(f"Hello, {name}! You are {age} years old.")
print("Data type of age:", type(age))
```

Exercise 2 — Arithmetic operators and temperature conversion

```
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9 / 5) + 32

print(f"{celsius}°C is equal to {fahrenheit}°F")
```

Exercise 3 — Integer division and remainder

```
N = int(input("Enter total number of notebooks (N): "))
S = int(input("Enter total number of students (S): "))

per_student = N // S
remainder = N % S

print(f"Each student receives: {per_student} notebook(s)")
print(f"Notebooks remaining: {remainder}")
```

Exercise 4 — Conditional statements

```
mark = float(input("Enter student mark (out of 20): "))

if mark >= 10:
    result = "Pass"
elif mark >= 8:
    result = "Resit"
else:
    result = "Fail"

print(f"Mark: {mark}/20 -> Result: {result}")
```

Exercise 6 — for loop and summation

```
n = int(input("Enter a positive integer n: "))
total_sum = 0

for i in range(1, n + 1):
    total_sum += i

print(f"The sum of all integers from 1 to {n} is: {total_sum}")
```

Exercise 7 — while loop countdown

```
count = int(input("Enter a positive integer for countdown: "))

while count > 0:
    print(count)
    count -= 1

print("0")
```

Exercise 8 — User-defined function and return

```
def calculate_average(m1, m2, m3):
    return (m1 + m2 + m3) / 3.0

mark1, mark2, mark3 = 12, 14.5, 16.5
avg = calculate_average(mark1, mark2, mark3)

print(f"Marks: {mark1}, {mark2}, {mark3}")
print(f"Calculated Average: {avg:.2f}")
```

Exercise 10 — math module and file handling

```
import math

number = 90.0
root_val = math.sqrt(number)
result_text = f"The square root of {number} is {root_val}"

filename = "result.txt"

with open(filename, "w") as file:
    file.write(result_text)

with open(filename, "r") as file:
    content = file.read()

print("File Content Readback:")
print(content)
```
